# Данный файл нужен для анализа заблокированных lsistings, по какой причине они блокнуты, как это связать с другими признаками

```python
import pandas as pd
from datetime import timedelta
```

```python
df = pd.read_csv("data/listings.csv")
delta_time = pd.to_timedelta(df['closed_at'].apply(lambda x: pd.to_datetime(x).timestamp()) - df['published_at'].apply(lambda x: pd.to_datetime(x).timestamp()), unit='s')
df["delta_time"] = delta_time
df = df.sort_values("delta_time", ascending=True)
df = df.sort_values("messages_total", ascending=True)
df = df.sort_values("views_total", ascending=True)



df_block = df[df["close_type"] == "заблокировано"]
df_block.shape
```

```python
from pathlib import Path

output_path = Path("data_temp/listing_id_of_block.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
df_block["listing_id"].to_csv(output_path, index=False)
```
